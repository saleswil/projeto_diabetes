```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

from sklearn.linear_model import LinearRegression #Biblioteca de regressão linear
from sklearn.metrics import (
    mean_absolute_error, 
    mean_squared_error,  
    PredictionErrorDisplay,
    root_mean_squared_error,
    r2_score # Métricas
)

from sklearn.model_selection import train_test_split #Trainar e testar o modelo

from src.auxiliares import dataframe_coeficientes
from src.config import DADOS_TRATADOS
from src.graficos import plot_coeficientes, plot_residuos
#scripts para melhorar a eficiência.


sns.set_theme(palette="bright")

RANDOM_STATE = 42 #Dessa forma, teremos os mesmos resultados.
```

```python
df = pd.read_parquet(DADOS_TRATADOS)

df.head()
```

```python
df.info()
```

```python
X = df.drop(columns="target") #Excluir a coluna alvo
y = df["target"] #Adicionar a coluna alvo

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=RANDOM_STATE
)

# Parâmetros para iniciar o treino e teste, vamos usar 20% dos dados para teste.
```

Acima, vemos que os dados foram embaralhados e que a maior parte deles foi destinada para o treino.

```python
regressor = LinearRegression() #regressor/reg é o nome usal do nosso modelo 

regressor.fit(X_train, y_train)
```

```python
regressor.score(X_train, y_train)
```

```python
regressor.score(X_test, y_test)
```

```python
regressor.coef_
```

```python
regressor.intercept_
```

```python
coefs = pd.DataFrame(
    data=regressor.coef_,
    index=X.columns,
    columns=["coeficiente"],
).sort_values(by="coeficiente")

coefs

#Visualização melhor do regressor.coef_
```

```python
coefs.plot.barh(); #Gráfico de barra horizontal

plt.axvline(x=0, color="0.5") #Adicionando uma linha horizontal
plt.xlabel("coeficiente") # Rótulo

plt.gca().get_legend().remove() # Remover a legenda nativa

plt.show()
```

```python
coefs = dataframe_coeficientes(regressor.coef_, X.columns)

coefs
```

Explicando o resultado do gráfico: a princípio, os triglicerídeos são o fator que mais impacta o avanço da diabetes.

No entanto, lá na etapa de EDA, observamos que os dados apresentam discrepâncias nas faixas de valores — por exemplo, variando de 0 a 1 ou de 1 a 1000.

```python
y_pred = regressor.predict(X_test)

y_pred
```

```python
r2_score(y_test, y_pred) #y_test e y_pred, valor real e valor previsto
```

```python
mean_absolute_error(y_test, y_pred)
```

```python
mean_squared_error(y_test, y_pred)
```

```python
root_mean_squared_error(y_test, y_pred)
```

Métricas do valor previsto.

```python
PredictionErrorDisplay.from_predictions(
    y_true=y_test,
    y_pred=y_pred,
);

#Visualizar a previsão de erro do modelo. Somente resíduos.
```

```python
PredictionErrorDisplay.from_predictions(
    y_true=y_test,
    y_pred=y_pred,
    kind="actual_vs_predicted"
);
```

```python
residuos = y_test - y_pred

sns.histplot(residuos, kde=True);
```

A maior concentração dos dados próxima à reta de previsão é um bom sinal, pois indica que o modelo está fazendo previsões razoavelmente precisas para a maior parte dos casos. Quanto mais os pontos se aproximam da reta (linha diagonal), melhor a performance do modelo, já que isso significa que as previsões estão mais próximas dos valores reais.

```python
plot_residuos(y_test, y_pred) # Valor real vs valor previsto
```
